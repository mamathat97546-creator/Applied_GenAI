loop with conditional statements

```python
# print the values from 1 to 10 without 5
```

```python
# 1,2,3,4,6,7,8,9,10
```

```python
for i in range(1,11):
  # print(i)
  if i==5:
    continue
  print(i)
```

```python
for i in range(1,11):
  # print(i)
  if i==5:
    continue
  else:
    print(i)
```

```python
for i in range(1,11):
  # print(i)
  if i==5:
    break
  else:
    print(i)
```
