**if-elif-else ladder**

create a ranking system-A(90 tO 100),B(75 to 89),C(60 to 74),D(50 to 59),E(35 to 49),F(0 to 34)

```python
score=73
```

```python
if score>=90:
  print('A')
elif score>=75 and score<90:
  print('B')
elif score>=60 and score<75:
  print('C')
elif score>=50 and score<60:
  print('D')
elif score>=35 and score>50:
  print('E')
else:
  print('F')
```
